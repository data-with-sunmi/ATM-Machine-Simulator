# ATM Machine Simulator

This notebook simulates a simple ATM system using Python.
It allows users to check balance, deposit, withdraw money and exit the system.


```python
def atm_machine():
    balance = 10000

    while True:
        print("\nWelcome to Sunmi's ATM Machine! What would you like to do today?")

        print("1. Check Balance") # to check your balance

        print("2. Deposit Money") # to deposit 

        print("3. Withdraw Money")# to withdraw 

        print("4. Exit") # to exit the ATM machine


        choice = input("Enter your choice please: ") 

        if choice == "1":  #Press 1 to check your balance (This function simply displays the users current account balance)

            print("Your balance is:", balance)

        elif choice == "2":  #Press 2 to deposit(This function allows users to add money to their account,the input amount is validated and added to the balance)
                
            try:
                amount = float(input("Kindly enter any amount to deposit: ")) 
                balance += amount
                print("Your deposit was successful! Your new balance is:", balance)
            except ValueError:
                print("Invalid input, kindly enter a number.")

        elif choice == "3": #Press 3 to withdraw(This function lets users withdraw funds,it checks whether the balance is sufficient before processing the withdrawal.If not, it shows an “insufficient balance” message.)
            try:
                amount = float(input("Enter the amount you want to withdraw: "))
                if amount <= balance:
                    balance -= amount
                    print("Withdrawal successful! Your new balance is:", balance)
                else:
                    print("Insufficient balance, please try again later.")
            except ValueError:
                print("Invalid input, kindly enter a number.")

        elif choice == "4": #Press 4 to exit(When the user chooses to exit, this function ends the session gracefully with a goodbye message)
            print("Thank you for banking with us. Have a nice day!")
            break

        else:
            print("Invalid choice, please try again.")


atm_machine()
```
